```python
# Task 2: Mark the success/failed launches for each site on the map
from folium.plugins import MarkerCluster
# 1. Initialize the map (reusing your existing map setup)
site_map = folium.Map(location=nasa_coordinate, zoom_start=5)

# [Re-insert your Task 1 code here to keep the launch site circles/labels if desired]

# 2. Create a MarkerCluster object for the launch outcomes
marker_cluster = MarkerCluster()

# 3. Iterate through the main dataframe (spacex_df) to mark each launch
for index, row in spacex_df.iterrows():
    coordinate = [row['Lat'], row['Long']]
    
    # Determine icon color based on launch success
    # Assuming 1 = Success, 0 = Failure
    if row['class'] == 1:
        icon_color = 'green'
        label = f"{row['Launch Site']} - Success"
    else:
        icon_color = 'red'
        label = f"{row['Launch Site']} - Failed"
        
    # Create a standard Marker with the designated color
    marker = folium.Marker(
        location=coordinate,
        popup=label,
        icon=folium.Icon(color='white', icon_color=icon_color)
    )
    
    # Add the marker to the cluster instead of directly to the map
    marker_cluster.add_child(marker)

# 4. Add the cluster layer to the map
site_map.add_child(marker_cluster)

# Display the map
site_map
```

*TODO:* Create a new column in `spacex_df` dataframe called `marker_color` to store the marker colors based on the `class` value


```python

# Apply a function to check the value of `class` column
# If class=1, marker_color value will be green
# If class=0, marker_color value will be red
# 1. Define a function to check the value of the class column
def assign_marker_color(launch_class):
    if launch_class == 1:
        return 'green'
    else:
        return 'red'

# 2. Apply the function to the 'class' column and assign it to 'marker_color'
spacex_df['marker_color'] = spacex_df['class'].apply(assign_marker_color)

# 3. View the tail of the dataframe to verify the new column was created successfully
spacex_df.tail(5)
```

Next, we need to explore and analyze the proximities of launch sites.


Let's first add a `MousePosition` on the map to get coordinate for a mouse over a point on the map. As such, while you are exploring the map, you can easily find the coordinates of any points of interests (such as railway)


```python
# Add Mouse Position to get the coordinate (Lat, Long) for a mouse over on the map
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)

site_map.add_child(mouse_position)
site_map
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
# find coordinate of the closet coastline
# e.g.,: Lat: 28.56367  Lon: -80.57163
# distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
# 1. Define the Launch Site coordinates (Example: CCAFS SLC-40)
launch_site_lat = 28.562302
launch_site_lon = -80.577356

# 2. Define the closest coastline coordinate (obtained from inspecting the map)
coastline_lat = 28.56367
coastline_lon = -80.57163

# 3. Calculate the distance using your function
distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)

# Print the result
print(f"Distance to the closest coastline: {distance_coastline:.2f} km")
```

```python
# Create and add a folium.Marker on your selected closest coastline point on the map
# Display the distance between coastline point and launch site using the icon property 
# for example
# distance_marker = folium.Marker(
#    coordinate,
#    icon=DivIcon(
#        icon_size=(20,20),
#        icon_anchor=(0,0),
#        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance),
#        )
#    )
# 1. Define the coordinate for your selected coastline point
coastline_coordinate = [coastline_lat, coastline_lon]

distance_marker = folium.Marker(
    location=coastline_coordinate,
    icon=DivIcon(
        icon_size=(20, 20),
        icon_anchor=(0, 0),
        html='<div style="font-size: 12pt; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance_coastline),
    )
)


site_map.add_child(distance_marker)

site_map
```

*TODO:* Draw a `PolyLine` between a launch site to the selected coastline point


```python

launch_site_lat = 28.563197
launch_site_lon = -80.576820
launch_coord = [launch_site_lat, launch_site_lon]

proximities = {
    "Coastline": [28.56367, -80.57163],
    "Highway": [28.56335, -80.57085],
    "Railway": [28.57206, -80.58525],
    "City (Titusville)": [28.61200, -80.80788]
}
for name, prox_coord in proximities.items():

    distance = calculate_distance(launch_site_lat, launch_site_lon, prox_coord[0], prox_coord[1])
    
 
    distance_marker = folium.Marker(
        location=prox_coord,
        icon=DivIcon(
            icon_size=(20, 20),
            icon_anchor=(0, 0),
            html=f'<div style="font-size: 12pt; color:#d35400;"><b>{distance:.2f} KM</b></div>',
        )
    )
    site_map.add_child(distance_marker)
    
    # Create and add the trace connection path line
    line = folium.PolyLine(
        locations=[launch_coord, prox_coord],
        weight=1,
        color='blue' if "City" not in name else 'red' # Highlight the city line uniquely
    )
    site_map.add_child(line)


site_map
```
